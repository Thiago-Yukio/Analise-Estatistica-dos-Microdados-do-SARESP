```python
lista_serie_performance_historico=[]
cont=2011
for i in list_proficiencia_saresp:
    lista_serie_performance=[]
    series = ['5º Ano EF', '7º Ano EF', '9º Ano EF','EM-3ª série']
    disciplinas = ['LÍNGUA PORTUGUESA','MATEMÁTICA','HISTÓRIA','GEOGRAFIA']
    lista_serie_performance.append(cont)
    for serie in series:
        serie_performance = {'SERIE_ANO': serie}
        for disciplina in disciplinas:
            if 'DS_COMP' in i.columns:
                notas = i.loc[
                (i['SERIE_ANO'] == serie) &
                (i['DS_COMP'] == disciplina),
                'MEDPROF'
                ]
            if 'ds_comp' in i.columns:
                notas = i.loc[
                (i['SERIE_ANO'] == serie) &
                (i['ds_comp'] == disciplina),
                'medprof'
                ]
            notas = pd.to_numeric(notas, errors='coerce')
            media=round(notas.mean(),2)
            serie_performance[disciplina] =(0 if pd.isna(media) else float(round(media, 2)))

        lista_serie_performance.append(serie_performance)
    if cont==2019:
        cont+=1
    cont+=1
    lista_serie_performance_historico.append(lista_serie_performance.copy())

```

```python
hist=pd.DataFrame(lista_serie_performance_historico,columns=['Ano','Histórico do 5° ano','Histórico do 7° ano','Histórico do 9° ano','Histórico do EM-3° série'])
```

```python
lista_serie_performance=[]

series = ['5º Ano EF', '7º Ano EF', '9º Ano EF','EM-3ª série']
disciplinas = ['LÍNGUA PORTUGUESA','MATEMÁTICA','HISTÓRIA','GEOGRAFIA']

for serie in series:
    serie_performance = {'SERIE_ANO': serie}
    for disciplina in disciplinas:
        notas = df_proficiencia_serasp_2011.loc[
            (df_proficiencia_serasp_2011['SERIE_ANO'] == serie) &
            (df_proficiencia_serasp_2011['DS_COMP'] == disciplina),
            'MEDPROF'
        ]
        notas = pd.to_numeric(notas, errors='coerce')
        media=round(notas.mean(),2)
        serie_performance[disciplina] =(0 if pd.isna(media) else float(round(media, 2)))

    lista_serie_performance.append(serie_performance)

print(lista_serie_performance)
```

```python
lista_df = []

coluna_historico = ['Histórico do 5° ano','Histórico do 7° ano','Histórico do 9° ano','Histórico do EM-3° série']

for coluna in coluna_historico:
    lista = []

    for ano, historico in zip(hist['Ano'], hist[coluna]):

        if not isinstance(historico, dict):
            continue

        a = historico.get('LÍNGUA PORTUGUESA')
        b = historico.get('MATEMÁTICA')

        if pd.isna(a) or pd.isna(b) or a == 0 or b == 0:
            continue

        lista.append({
            'ANO': ano,
            'LÍNGUA PORTUGUESA': a,
            'MATEMÁTICA': b,
        })

    df_historico_serie = pd.DataFrame(lista)
    lista_df.append(df_historico_serie)
```

```python
for i,a in zip(lista_df,coluna_historico):
    i.plot(kind='line',x='ANO',ylabel='NOTA',title=f'Histórico da nota do {a}')
    #plt.savefig(f"Histórico da nota do {a}.png", dpi=300, bbox_inches="tight")
```

```python
import pandas as pd

list_df_periodo_hist = []

for ano, i in zip(hist['Ano'], list_proficiencia_saresp):
    x1 = i.get('PERIODO')

    if x1 is None:
        continue

    contagem = x1.value_counts()
    for periodo, quantidade in contagem.items():
        list_df_periodo_hist.append({
            'ANO': ano,
            'PERIODO': periodo,
            'QUANTIDADE': quantidade
        })

df_periodo_hist = pd.DataFrame(list_df_periodo_hist)

display(df_periodo_hist)
```
