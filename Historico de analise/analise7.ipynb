```python
cont = 2011
lista_df = []
for df in list_proficiencia_saresp:
    if cont == 2020:
        cont += 1
    dict_df = {}
    for c in df:
        if c.upper().strip() == 'PERIODO':
            dict_df['PERIODO'] = df.get(c, default='Não encontrado').copy()
        if c.upper().strip() == 'SERIE_ANO':
            dict_df['SERIE_ANO'] = df.get(c, default='Não encontrado').copy()
        if c.upper().strip() == 'NOMEDEPBOL':
            dict_df['NOMEDEPBOL'] = df.get(c, default='Não encontrado').copy()
    dict_df['ANO'] = cont
    lista_df.append(pd.DataFrame(dict_df))
    cont += 1
df_a = pd.concat(lista_df, ignore_index=True)

df_a
```

```python
df_a['SERIE_ANO'].value_counts()
```

```python
df_a['SERIE_ANO']=df_a['SERIE_ANO'].replace('EM-3Âª sÃ©rie','EM-3ª série')
df_a['SERIE_ANO']=df_a['SERIE_ANO'].replace('7Âº Ano EF','7º Ano EF')
df_a['SERIE_ANO']=df_a['SERIE_ANO'].replace('5Âº Ano EF','5º Ano EF')
df_a['SERIE_ANO']=df_a['SERIE_ANO'].replace('9Âº Ano EF','9º Ano EF')
df_a['SERIE_ANO'].value_counts()
```

```python
df_a = df_a[~df_a['SERIE_ANO'].isin(['2º Ano EF', '3º Ano EF'])]
```

```python
series = ['9º Ano EF', '5º Ano EF', 'EM-3ª série', '7º Ano EF']

contagens = df_a['SERIE_ANO'].value_counts().reindex(series, fill_value=0)

plt.figure(figsize=(6, 7))

barras = plt.bar(series, contagens.values,color='orange')

plt.bar_label(barras, padding=3, fmt='%d')

plt.title('Distribuição dos Dados por Série entre (2011–2025)')
plt.xlabel('Série/Ano')
plt.ylabel('Quantidade')
plt.xticks(rotation=45)
plt.savefig('Distribuição dos Dados por Série entre (2011–2025).png', dpi=300, bbox_inches='tight')
plt.tight_layout()
plt.show()
```

```python
df_a['PERIODO']=df_a['PERIODO'].replace('MANHA','MANHÃ')
df_a['PERIODO']=df_a['PERIODO'].replace('MANHÃ','Diurno')
df_a['PERIODO']=df_a['PERIODO'].replace('TARDE','Vespertino')
df_a['PERIODO']=df_a['PERIODO'].replace('NOITE','Noturno')
df_a['PERIODO']=df_a['PERIODO'].replace('GERAL','Ensino Integral')


contagens = df_a.groupby(['ANO', 'PERIODO']).size().unstack(fill_value=0)

plt.figure(figsize=(8, 4))

for periodo in contagens.columns:
    plt.plot(
        contagens.index,
        contagens[periodo],
        marker='o',
        label=periodo
    )

plt.title('Quantidade de Registros por Ano e Período (2011–2025)')
plt.xlabel('Ano')
plt.ylabel('Quantidade')
plt.legend(title='Período',loc='center left',bbox_to_anchor=(1, 0.5))
plt.grid(True, alpha=0.3)
plt.savefig('Quantidade de Registros por Ano e Período (2011–2025).png', dpi=300, bbox_inches='tight')
plt.tight_layout()
plt.show()
```
