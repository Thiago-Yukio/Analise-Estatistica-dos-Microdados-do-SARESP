```python
dados_auxiliares = []

for ano, df in zip(range(2011, 2026), lista_fluxo_dfs):

    df = df.copy()
    df.columns = df.columns.str.upper().str.strip()

    for coluna in ['APR_3', 'REP_3']:

        df[coluna] = (
            df[coluna]
            .astype(str)
            .str.strip()
            .str.replace(',', '.', regex=False)
            .replace('-', pd.NA)
        )

        df[coluna] = pd.to_numeric(df[coluna], errors='coerce')

    dados_auxiliares.append(pd.DataFrame({
        'ANO': ano,
        'APR_3': df['APR_3'],
        'REP_3': df['REP_3']
    }))

df_tratado = pd.concat(dados_auxiliares, ignore_index=True)

df_tratado = df_tratado.dropna(subset=['APR_3', 'REP_3']).reset_index(drop=True)
```

```python
dados_auxiliares = []

for ano, df in zip(range(2011, 2026), lista_fluxo_dfs):

    df = df.copy()
    df.columns = df.columns.str.upper().str.strip()

    for coluna in ['APR_3', 'REP_3']:

        df[coluna] = (
            df[coluna]
            .astype(str)
            .str.strip()
            .str.replace(',', '.', regex=False)
            .replace('-', pd.NA)
        )

        df[coluna] = pd.to_numeric(df[coluna], errors='coerce')

    # Tratamento específico de 2014
    if ano == 2014:

        # Identificar linhas em que ambas as variáveis estão ausentes
        ausentes = df.index[
            df['APR_3'].isna() & df['REP_3'].isna()
        ]

        # Selecionar aleatoriamente 1500 registros
        selecionados = ausentes.to_series().sample(
            n=1500,
            random_state=42
        )

        # Preencher com a mediana de cada variável
        df.loc[selecionados, 'APR_3'] = df['APR_3'].median()
        df.loc[selecionados, 'REP_3'] = df['REP_3'].median()

    dados_auxiliares.append(pd.DataFrame({
        'ANO': ano,
        'APR_3': df['APR_3'],
        'REP_3': df['REP_3']
    }))

df_tratado = pd.concat(dados_auxiliares, ignore_index=True)

# Remover os demais nulos
df_tratado = df_tratado.dropna(
    subset=['APR_3', 'REP_3']
).reset_index(drop=True)
```

```python
df_tratado = df_tratado[
    (df_tratado['REP_3'] >= 0) &
    (df_tratado['REP_3'] <= 100)
].reset_index(drop=True)
```

```python
estatisticas_apr = df_tratado.groupby('ANO')['REP_3'].agg(
    media='mean',
    mediana='median',
    moda=lambda x: x.mode().iloc[0] if not x.mode().empty else None,
    desvio_padrao='std'
).to_dict('index')

df_estatistica = pd.DataFrame.from_dict(
    estatisticas_apr,
    orient='index'
).reset_index(names='ANO')
```

```python
plt.figure(figsize=(12, 6))

df_estatistica.plot(
    x='ANO',
    y=['media', 'mediana', 'desvio_padrao'],
    kind='line',
    marker='o',
    figsize=(12, 6)
)

plt.title('Evolução das Estatísticas de Reprovação Escolar no Ensino Médio (2011–2025)')
plt.xlabel('Ano')
plt.ylabel('Porcentagem (%)')
plt.xticks(df_estatistica['ANO'])
plt.grid(alpha=0.3)
plt.legend(['Média', 'Mediana', 'Desvio padrão'])
plt.tight_layout()

plt.show()
```

```python
dados_auxiliares = []

for ano, df in zip(range(2011, 2026), lista_fluxo_dfs):

    df = df.copy()
    df.columns = df.columns.str.upper().str.strip()

    for coluna in ['APR_2', 'REP_2']:
        df[coluna] = (
            df[coluna]
            .astype(str)
            .str.strip()
            .str.replace(',', '.', regex=False)
            .replace('-', pd.NA)
        )

        df[coluna] = pd.to_numeric(df[coluna], errors='coerce')

    # Tratamento específico de 2014
    if ano == 2014:
        ausentes = df.index[
            df['APR_2'].isna() & df['REP_2'].isna()
            ]

        selecionados = ausentes.to_series().sample(
            n=1500,
            random_state=42
        )

        df.loc[selecionados, 'APR_2'] = df['APR_2'].median()
        df.loc[selecionados, 'REP_2'] = df['REP_2'].median()

    dados_auxiliares.append(pd.DataFrame({
        'ANO': ano,
        'APR_2': df['APR_2'],
        'REP_2': df['REP_2']
    }))

df_tratado_2 = pd.concat(dados_auxiliares, ignore_index=True)

# Remover valores nulos
df_tratado_2 = df_tratado_2.dropna(
    subset=['APR_2', 'REP_2']
).reset_index(drop=True)

# Remover valores fora da escala percentual
df_tratado_2 = df_tratado_2[
    (df_tratado_2['APR_2'].between(0, 100)) &
    (df_tratado_2['REP_2'].between(0, 100))
    ].reset_index(drop=True)

# Calcular estatísticas de REP_2
estatisticas_rep_2 = df_tratado_2.groupby('ANO')['REP_2'].agg(
    media='mean',
    mediana='median',
    moda=lambda x: x.mode().iloc[0] if not x.mode().empty else None,
    desvio_padrao='std'
).to_dict('index')

df_estatistica_2 = pd.DataFrame.from_dict(
    estatisticas_rep_2,
    orient='index'
).reset_index(names='ANO')

plt.figure(figsize=(12, 6))

df_estatistica_2.plot(
    x='ANO',
    y=['media', 'mediana', 'desvio_padrao'],
    kind='line',
    marker='o',
    figsize=(12, 6)
)

plt.title('Porcentagem de Alunos Reprovados no Ensino Fundamental - Anos Finais no ensino fundamental do 9° ano (2011–2025)')
plt.xlabel('Ano')
plt.ylabel('Porcentagem (%)')
plt.xticks(df_estatistica_2['ANO'])
plt.grid(alpha=0.3)
plt.legend(['Média', 'Mediana', 'Desvio padrão'])
plt.tight_layout()

plt.show()
```
