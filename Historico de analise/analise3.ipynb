```python
contagem = df_historico_matricula_turma_21_23['GrupoEnsino'].value_counts()

limite = contagem.sum() * 0.02

principais = contagem[contagem >= limite].copy()
outros = contagem[contagem < limite].sum()

if outros > 0:
    principais['Outros'] = outros

total = principais.sum()

indice_fatia = iter(principais.index)

def autopct_personalizado(pct):
    grupo = next(indice_fatia)

    if grupo == 'Outros':
        return ''

    return f'{pct:.1f}%'.replace('.', ',')

fig, ax = plt.subplots(figsize=(10, 6))

wedges, _, autotexts = ax.pie(
    principais,
    startangle=90,
    labels=None,
    autopct=autopct_personalizado,
    pctdistance=0.65,
    textprops={
        'fontsize': 9,
        'fontweight': 'bold'
    }
)

for texto in autotexts:
    texto.set_color('white')

legenda = [
    f"{grupo}: {valor:,.0f} ({valor / total * 100:.1f}%)"
    .replace(',', 'X').replace('.', ',').replace('X', '.')
    for grupo, valor in principais.items()
]

ax.legend(
    wedges,
    legenda,
    title='Grupo de Ensino (matrículas e %)',
    loc='center left',
    bbox_to_anchor=(1, 0.5),
    frameon=False,
    fontsize=9,
    labelspacing=1.2
)

ax.set_title(
    'Matrículas por Grande Grupo de Ensino entre 2021 e 2023'
)

ax.axis('equal')
plt.tight_layout()

plt.savefig(
    'matriculas_grupo_ensino_2021_2023.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```
